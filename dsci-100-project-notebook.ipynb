- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).

```r
missing_na <- wmm_data |>
    summarize(across(year:total_wins_pct, ~ sum(is.na(.x)))) |>
    pivot_longer(
        cols = year:total_wins_pct,
        names_to = "column_name",
        values_to = "number_missing"
        )
missing_na
```

```r
### EDIT THIS CODE CELL FOR PLOT 1 ###
#plot 1 (histogram)
options(repr.plot.width = 8, repr.plot.height = 6)

wins_pct_hist <- ggplot(tidy_wmm, aes(x = total_wins_pct)) +
    geom_histogram(bins = 25, fill = "orange", colour = "white") +
    labs(
        title = "Distribution of Team's Total Win Percentage in Womens March Madness",
        x = "Total Win Percentage (%)",
        y = "Number of Teams"
    ) +
theme(text = element_text(size = 14))
wins_pct_hist
```

```r
### EDIT THIS CODE CELL FOR PLOT 2 ###
options(repr.plot.width = 6, repr.plot.height = 6) #first set the dimensions so that it is visually readable

number_of_obs <- nrow(tidy_wmm)

bid_prop <- tidy_wmm |>
    group_by(bid) |>
    summarize(proportion = n()/number_of_obs)
# I did this to find the proportion first before making the plot
bid_prop_plot <- ggplot(bid_prop, aes(x = bid, y = proportion, fill = bid)) +
    geom_bar(stat = "identity") +
    labs(
        title = "Proportions of At-large and Automatic Bids",
        x = "Type of Tournament Bid",
        y = "Relative Proportion of Teams",
        fill = "Type of Bid"
    ) +
scale_fill_manual(values = c("orange", "steelblue")) +
theme(text = element_text(size = 14))

bid_prop_plot
```

Visualization 1: %%% EDIT THIS MARKDOWN CELL%%%

The histogram that is stacked shows and compares the distribution of total win percent across both of the tournament qualification types. This is a categorical variable with two options, automatic and at-large. This is relevant to the research question because it shows the centres, ranges and spreads of the automatic and at-large qualifiers. By looking at the figures, we can see and determine how effective the total win percentage (total_win_pct) can distinguish at-large and auto bids in the KNN or K nearest neighbours classification model.

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 1###
options(repr.plot.width = 8, repr.plot.height = 6)

relationship_plot_1 <- ggplot(tidy_wmm, aes(x = total_wins_pct, fill = bid)) +
geom_histogram(bins = 25, colour = "black")+
facet_grid(rows = vars(bid)) +
labs(
    title = "Distribution of Total Wins by Bid Type",
    x = "Total Wins Percentage (%)",
    y = "Number of Teams",
    fill = "Type of Bid"
) +
scale_fill_manual(values = c("lightblue", "darkorange")) +
theme(text = element_text(size = 16))

relationship_plot_1
```

Visualization 2: %%% EDIT THIS MARKDOWN CELL%%%
This bar chart looks at the average total win percentages between the teams that received at-large and auto bids. It compares them side by side. This is relevant to the research because it summarizes and makes it easy to read whether a certain bid category has a higher average total win percentage than the other. With the first plot and second plot together, it helps us determine whether or not the total win percentage differs enough between both of the categories to ensure that it is a useful predictor of the bid type.

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 2###
options(repr.plot.width = 6, repr.plot.height = 6)

relationship_plot_2 <- tidy_wmm |>
    group_by(bid) |>
    summarize(mean_total_wins_pct = mean(total_wins_pct, na.rm = TRUE)) |>
    ggplot(aes(x = bid, y = mean_total_wins_pct, fill = bid)) +
    geom_bar(stat = "identity") +
    labs(
        title = "Mean Total Wins Percentage For At-large and Auto Bids",
        x = "Bid Type",
        y = "Mean Total Wins Percentage (%)",
        fill = "Bid Type") +
scale_fill_manual(values = c("blue", "orange")) +
theme(text = element_text(size = 12))

relationship_plot_2
```
